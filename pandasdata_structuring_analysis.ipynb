```python
import pandas as pd
data={
    'roll_no':[1,2,3,4,5],
    'name':['Nithin','Asha','Naveen','John','Neha'],
    'age':[20,21,22,20,23],
    'marks':[85,76,92,67,88]
}
# to structure data into table sudents
students=pd.DataFrame(data)

print('student table')
print(students)
print("\n a) top 2 rows")
print (students.head(2))

# print("\n b) last 2 rows")
# print (students.tail(2))

print("\n b) sudents with marks greater than 80")
print(students[students['marks']>80])

# print("\n c) students with age between 20 and 22")
# print(students[(students['age']>=20) & (students['age']<=22)])

names_sarting_with_N=students[students['name'].str.startswith('N')]
rmaining_names=students[~students['name'].str.startswith('N')]

print("\n c) students with name starting with N")
print(names_sarting_with_N)
print("\n d) remaining students")
print(rmaining_names)
```
