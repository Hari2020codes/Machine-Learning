```python
from sklearn.linear_model import LinearRegression
import numpy as np
X=np.array([[1], [2], [3], [4], [5]])
y =np.array([2, 4, 5, 4, 5])
#Train model
model = LinearRegression().fit(X, y)
#Predict and show results
print("Coefficient:", model.coef_[0])
print("Intercept:", model.intercept_)
print("Prediction for x=6:", model.predict([[6]])[0])
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn import datasets,linear_model
from sklearn import metrics
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import mean_squared_error,r2_score
df=pd.read_excel('aptitude.xlsx')

X=df[['x1']]
y=df[['y1']]
x_train,x_test,y_train,y_test=train_test_split(X,y,test_size=0.3)
clf=LinearRegression()
clf.fit(x_train,y_train)
y_pred=clf.predict(x_test)
print(y_pred)
print("y=",clf.intercept_,"+",clf.coef_,"* x")
newvalue=float(input("enter the score:"))
y_pred=clf.intercept_+clf.coef_*newvalue
print(y_pred)
```

To upload file to colab permanently
=====================================
files uploaded directly to the Colab environment using files.upload() are temporary and are deleted when the Colab runtime disconnects or restarts. This means you would need to re-upload them each time you start a new session or after a runtime restart.

If you want to avoid re-uploading files every time, you can consider these options:

Mount Google Drive: You can mount your Google Drive to Colab and store your files there. This way, your files persist across sessions, and you can access them directly from your Drive.
Use a cloud storage service: For larger or more permanent datasets, using a service like Google Cloud Storage or a public URL is often a better approach.
